# Given the dataset below, perform the required task for each question and display the output.



```python
transactions = [
    ['A', 'B'],   # T1
    ['A'],        # T2
    ['B'],        # T3
    ['A', 'B'],   # T4
    []            # T5 (neither A nor B)
]
transactions
```

# Q1- Perform Apirioi algorithm to get the frequent itemsets given min support = 3 and min conf 80%.

```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules

# Convert the list of transactions into a one-hot encoded DataFrame
te = TransactionEncoder()
te_ary = te.fit(transactions).transform(transactions)
df = pd.DataFrame(te_ary, columns=te.columns_)

# Calculate min_support as a percentage (3 / total number of transactions)
min_support_value = 3 / len(transactions)

# Apply Apriori algorithm to find frequent itemsets
frequent_itemsets = apriori(df, min_support=min_support_value, use_colnames=True)

print("Frequent Itemsets:")
print(frequent_itemsets)

# Generate association rules
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.8)

print("\nAssociation Rules (min confidence = 80%):")
print(rules)
```

# Q4- Returns a dict with:
      - counts
      - support
      - confidence (X->Y, Y->X)
      - lift (X->Y)
      - interest (same as lift via alternative formula)
      - contingency table counts (a,b,c,d)
      - h-confidence

```python
def analyze_rule(transactions, antecedent, consequent):
    """
    Calculates various metrics for an association rule (antecedent -> consequent).

    Args:
        transactions (list of lists): A list of transactions, where each transaction is a list of items.
        antecedent (tuple or frozenset): The items forming the antecedent (X) of the rule.
        consequent (tuple or frozenset): The items forming the consequent (Y) of the rule.

    Returns:
        dict: A dictionary containing the calculated metrics:
              - counts (antecedent, consequent, union)
              - support (antecedent, consequent, union)
              - confidence (X->Y, Y->X)
              - lift (X->Y)
              - interest (same as lift)
              - contingency_table (a, b, c, d)
              - h_confidence
    """
    num_transactions = len(transactions)
    if num_transactions == 0:
        return {}

    antecedent_set = frozenset(antecedent)
    consequent_set = frozenset(consequent)
    union_set = antecedent_set.union(consequent_set)

    # Calculate counts
    count_X = 0
    count_Y = 0
    count_X_union_Y = 0

    for t in transactions:
        t_set = frozenset(t)
        if antecedent_set.issubset(t_set):
            count_X += 1
        if consequent_set.issubset(t_set):
            count_Y += 1
        if union_set.issubset(t_set):
            count_X_union_Y += 1

    # Calculate support
    support_X = count_X / num_transactions
    support_Y = count_Y / num_transactions
    support_X_union_Y = count_X_union_Y / num_transactions

    # Calculate confidence
    confidence_X_to_Y = support_X_union_Y / support_X if support_X > 0 else 0.0
    confidence_Y_to_X = support_X_union_Y / support_Y if support_Y > 0 else 0.0

    # Calculate lift
    lift_X_to_Y = confidence_X_to_Y / support_Y if support_Y > 0 else 0.0
    interest_X_to_Y = lift_X_to_Y # As per prompt: "same as lift"

    # Calculate contingency table counts
    a = count_X_union_Y
    b = count_X - count_X_union_Y # X but not Y
    c = count_Y - count_X_union_Y # Y but not X
    d = num_transactions - (a + b + c) # Neither X nor Y

    # Calculate h-confidence
    h_confidence = min(confidence_X_to_Y, confidence_Y_to_X)

    result_dict = {
        "counts": {
            "antecedent": count_X,
            "consequent": count_Y,
            "union": count_X_union_Y
        },
        "support": {
            "antecedent": support_X,
            "consequent": support_Y,
            "union": support_X_union_Y
        },
        "confidence": {
            "X->Y": confidence_X_to_Y,
            "Y->X": confidence_Y_to_X
        },
        "lift": lift_X_to_Y,
        "interest": interest_X_to_Y,
        "contingency_table": {
            "a": a,
            "b": b,
            "c": c,
            "d": d
        },
        "h_confidence": h_confidence
    }

    return result_dict

# Example usage with the 'transactions' data
# Let's analyze the rule {A} -> {B}
r_A_to_B = analyze_rule(transactions, ('A',), ('B',))
print("\nMetrics for rule {A} -> {B}:")
print(r_A_to_B)

# Let's analyze the rule {B} -> {A}
r_B_to_A = analyze_rule(transactions, ('B',), ('A',))
print("\nMetrics for rule {B} -> {A}:")
print(r_B_to_A)

# Example with a rule that might not exist or have different stats
# r_A_to_empty = analyze_rule(transactions, ('A',), ()) # Empty consequent/antecedent might need special handling
# print("\nMetrics for rule {A} -> {}:")
# print(r_A_to_empty)
```
